```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
        self.down = None

    def __str__(self):
        return str(self.value)
```

```python
print(Node(4))
```

```python
class Stack:
    def __init__(self):
        self.head = None
        self.no_of_stacks = 0

    def push(self, value):
        new_node = Node(value)
        if self.head:
            new_node.down = self.head
            self.head.next = new_node
        self.head = new_node
        self.no_of_stacks += 1

    def pop(self):
        if self.is_empty():
            return None

        popped_node = self.head
        if self.no_of_stacks > 1:
            self.head = popped_node.down
            self.head.next = None
        else:
            self.head = None

        popped_node.down = None
        self.no_of_stacks -= 1
        return popped_node

    def peek(self, index):
        if index < 0 or index >= self.no_of_stacks:
            raise Exception(f'Index {index} Out of Range')

        current = self.head
        for _ in range(index):
            current = current.down
        return current

    def is_empty(self):
        return self.no_of_stacks == 0

    def delete(self):
        self.head = None
        self.no_of_stacks = 0

    def __str__(self):
        if self.is_empty():
            return "Stack is empty."

        lines = ["TOP"]
        current = self.head
        while current:
            lines.append(f" ↓\n[{current.value}]")
            current = current.down
        lines.append(" ↓\nBOTTOM")
        return "\n".join(lines)


```

```python
stacks = Stack()

stacks.push(3)
stacks.push(4)
stacks.push(5)

print(stacks.pop())
# print(stacks.delete())
print(stacks)
```
