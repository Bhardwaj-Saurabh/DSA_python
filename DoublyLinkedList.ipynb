```python
class Node:
    def __init__(self, value=None):
        self.value = value
        self.next = None
        self.prev = None

    def __str__(self):
        return str(self.value)

```

```python
class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.length = 0

    def prepend(self, value):
        new_node = Node(value)

        if self.length == 0:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node
        self.length += 1

    def append(self, value):
        new_node = Node(value)

        if self.length == 0:
            self.head = new_node
            self.tail = new_node
        else:
            new_node.prev = self.tail
            self.tail.next = new_node
            self.tail = new_node
        self.length += 1

    def insert(self, index, value):
        if index < 0 or index > self.length:
            raise Exception(f'Index {index} Out of Range.')

        new_node = Node(value)

        if index == 0:
            return self.prepend(value)
        elif index == self.length:
            return self.append(value)
        else:
            current = self.head
            for _ in range(index):
                current = current.next
            current.prev.next = new_node
            new_node.prev = current.prev
            new_node.next = current
            current.prev = new_node
            self.length += 1

    def get(self, index):
        if index < 0 or index >= self.length:
            raise Exception(f'Index {index} Out of Range.')

        current = self.head
        for _ in range(index):
            current = current.next

        return current

    def set(self, index, value):
        if index < 0 or index >= self.length:
            raise Exception(f'Index {index} Out of Range.')

        current = self.head
        for _ in range(index):
            current = current.next
        current.value = value

    def pop_first(self):

        if self.length == 0:
            return None

        current = self.head
        if self.length == 1:
            self.head = None
            self.tail = None
        else:
            self.head = current.next
            current.next = None
            self.head.prev = None
        self.length -= 1
        return current

    def pop(self):
        if self.length == 0:
            return None

        current = self.tail
        if self.length == 1:
            self.head = None
            self.tail = None
        else:
            self.tail = current.prev
            self.tail.next = None
            current.prev = None
        self.length -= 1
        return current

    def remove(self, index):
        if index < 0 or index >= self.length:
            raise Exception(f'Index {index} Out of Range.')


        if index == 0:
            return self.pop_first()
        elif index == self.length-1:
            return self.pop()
        else:
            current = self.head
            for _ in range(index):
                current = current.next

            prev = current.prev
            next = current.next
            prev.next = next
            current.prev = None
            current.next = None
            next.prev = prev
            self.length -= 1
        return current

    def empty(self):
        if self.length == 0:
            return None
        else:
            self.head = None
            self.tail = None


    def __str__(self):
        nodes = []
        current = self.head

        while current:
            nodes.append(f'{current.value}')
            current = current.next

        return " <==> ".join(nodes)

```
